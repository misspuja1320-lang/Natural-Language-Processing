```python
# Lowercasing

text=input("Enter the text:")
print("Original text:",text)
new_text=text.lower()
print("New text:",new_text)
```

```python
# Word Tokenization

import nltk
nltk.download('punkt')
nltk.download('punkt_tab')
from nltk.tokenize import word_tokenize
text=input("Enter the text:")
print("Original text:",text)
tokenize=word_tokenize(text)
print("Word tokenization:",tokenize)
```

```python
# Sentence Tokenization

import nltk
nltk.download('punkt')
nltk.download('punkt_tab')
from nltk.tokenize import sent_tokenize
text=input("Enter the text:")
print("Original text:",text)
tokenize=sent_tokenize(text)
print("Sentence tokenization:",tokenize)
```

```python
# Remove Punctuation

import string
text = input("Enter the text: ")
print("Original text:", text)
clean_text = text.translate(str.maketrans("", "", string.punctuation))
print("Text after removing punctuation:", clean_text)
```

```python
# Stop-word removal

import nltk
nltk.download('stopwords')
nltk.download('punkt')
nltk.download('punkt_tab')
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
text=input("Enter the text:")
print("Original text:",text)
word=word_tokenize(text.lower())
stop_word=set(stopwords.words('english'))
filtered_words=[]
for i in word:
    if i.isalpha() and i not in stop_word:
        filtered_words.append(i)
print("After stop-word removal:",filtered_words)
```

```python
# Stemming

import nltk
nltk.download('punkt')
nltk.download('punkt_tab')
from nltk.tokenize import word_tokenize
from nltk.stem import PorterStemmer
text=input("Enter the text:")
print("Original text:",text)
word=word_tokenize(text.lower())
stemmer=PorterStemmer()
print("Word\t\tStem")
print("-------------------")
for i in word:
    if i.isalpha():
        stem=stemmer.stem(i)
        print(i,"\t\t",stem)
```

```python
# Lemmatization

import nltk
nltk.download('wordnet')
nltk.download('omw-1.4')
from nltk.stem import WordNetLemmatizer
lemmatizer = WordNetLemmatizer()
words = ["running", "studies", "better", "playing", "cars"]
print("Word\t\tLemma")
print("-------------------------")
for word in words:
    lemma = lemmatizer.lemmatize(word)
    print(word, "\t\t", lemma)
```

```python
# Lemmatization with POS

import nltk
nltk.download('wordnet')
nltk.download('omw-1.4')
from nltk.stem import WordNetLemmatizer
lemmatizer = WordNetLemmatizer()
words = [
    ("running", "v"),
    ("studies", "v"),
    ("better", "a"),
    ("playing", "v"),
    ("cars", "n")
]
print("Word\t\tLemma")
print("-------------------------")
for word, pos in words:
    lemma = lemmatizer.lemmatize(word, pos=pos)
    print(word, "\t\t", lemma)
```
