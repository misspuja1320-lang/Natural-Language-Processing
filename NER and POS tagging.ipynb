```python
import spacy
nlp = spacy.load("en_core_web_sm")
text = input("Enter the text: ")
doc = nlp(text)
print("Entity\t\tType")
print("-------------------")
for ent in doc.ents:
    print(ent.text,  ent.label_)
```

```python
import nltk
nltk.download("punkt")
nltk.download("averaged_perceptron_tagger_eng")
text = input("Enter the text: ")
word=nltk.word_tokenize(text)
pos_tag=nltk.pos_tag(word)
print("Word\t\tPOS_tag")
print("---------------------")
for word,tag in pos_tag:
    print(f"{word}\t\t{tag}")
```

```python
import spacy
nlp = spacy.load("en_core_web_sm")
text = input("Enter the text: ")
doc = nlp(text)
print("------POS tag-----")
print("Word\t\tPOS_tag")
for token in doc:
    print(token.text,"\t\t",token.pos_)
print("------Named entity-----")
print("Entity\t\tLabel")
for ent in doc.ents:
    print(ent.text,"\t\t",ent.label_)
```

```python
import spacy
from collections import Counter
nlp = spacy.load("en_core_web_sm")
text = input("Enter the text: ")
doc = nlp(text)
pos_count=Counter()
for token in doc:
    if not token.is_punct:
        pos_count[token.pos_]+=1
print("POS tag\t\tCount")
print("--------------------")
for tag,count in pos_count.items():
    print(tag,"\t\t",count)
```
