```python
from nltk.sentiment import SentimentIntensityAnalyzer
import nltk
sia=SentimentIntensityAnalyzer()
t=input("Enter the text:")
score=sia.polarity_scores(t)
print("\n Sentiment scores:",score)
if score['compound']>=0.05:
    print("Sentiment:positive")
elif score['compound']<=0.05:
    print("Sentiment:nagative")
else:
    print("Sentiment:natural")
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
s=['I love this movie','I really enjoyed the movie','I hate this movie','I dislike this product']
labels=['positive','positive','nagative','nagative']
vectorizer=TfidfVectorizer()
x=vectorizer.fit_transform(s)
model=LogisticRegression()
model.fit(x,labels)
t=input("Enter the sentence:")
text_vector=vectorizer.transform([t])
prediction=model.predict(text_vector)
print("Predicted sentiment:",prediction[0])
```

```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
reviews=['I love this movie','I really enjoyed the movie','The movie is excellent','The movie is good',
         'I hate this movie','I dislike this product','Very bad movie','The acting is poor']
sentiments=['positive','positive','positive','positive','nagative','nagative','nagative','nagative']
vectorizer=CountVectorizer()
x=vectorizer.fit_transform(reviews)
model=MultinomialNB()
model.fit(x,sentiments)
review=input("Enter the reviews:")
review_vector=vectorizer.transform([review])
prediction=model.predict(review_vector)
print("Predicted sentiment:",prediction[0])
```
