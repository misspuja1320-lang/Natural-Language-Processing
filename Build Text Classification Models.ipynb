```python
#  Naive Bayes Text Classification

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
# Training data
texts = [
    "Win a free prize now",
    "Congratulations you won a lottery",
    "Claim your free reward",
    "Meeting is scheduled tomorrow",
    "Please submit the project report",
    "Class will start at 10 AM"
]
labels = [
    "Spam",
    "Spam",
    "Spam",
    "Ham",
    "Ham",
    "Ham"
]
# Convert text into TF-IDF features
vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(texts)
# Create and train model
model = MultinomialNB()
model.fit(X, labels)
# Test data
test_text = ["Congratulations! You won a free prize"]
test_X = vectorizer.transform(test_text)
# Prediction
prediction = model.predict(test_X)
print("Test Text:", test_text[0])
print("Prediction:", prediction[0])
```

```python
# Logistic Regression Text Classification

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
texts = [
    "I love this movie",
    "This movie is excellent",
    "The movie was fantastic",
    "Amazing movie and great acting",
    "I hate this movie",
    "This movie is terrible",
    "The movie was boring",
    "Worst movie ever"
]
labels = [
    "Positive",
    "Positive",
    "Positive",
    "Positive",
    "Negative",
    "Negative",
    "Negative",
    "Negative"
]
# Convert text into numerical features
vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(texts)
# Train Logistic Regression model
model = LogisticRegression()
model.fit(X, labels)
# Test sentence
test_text = ["The movie was fantastic"]
test_X = vectorizer.transform(test_text)
# Prediction
prediction = model.predict(test_X)
print("Test Text:", test_text[0])
print("Sentiment:", prediction[0])
```

```python
# Support Vector Machine (SVM) Text Classification

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.svm import SVC
texts = [
    "I like this product",
    "This product is excellent",
    "Very good product",
    "Amazing product",
    "I dislike this product",
    "This product is bad",
    "Very poor product",
    "Worst product"
]
labels = [
    "Positive",
    "Positive",
    "Positive",
    "Positive",
    "Negative",
    "Negative",
    "Negative",
    "Negative"
]
# TF-IDF feature extraction
vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(texts)
# Create SVM model
model = SVC(kernel='linear')
# Train model
model.fit(X, labels)
# Test text
test_text = ["This is a very good product"]
test_X = vectorizer.transform(test_text)
# Prediction
prediction = model.predict(test_X)
print("Test Text:", test_text[0])
print("Prediction:", prediction[0])
```

```python
# Train-Test Split for Text Classification

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score
texts = [
    "I love this product",
    "This product is excellent",
    "Very good quality",
    "Amazing product",
    "Fantastic product",
    "I hate this product",
    "This product is terrible",
    "Very bad quality",
    "Worst product",
    "Poor quality product"
]
labels = [
    "Positive",
    "Positive",
    "Positive",
    "Positive",
    "Positive",
    "Negative",
    "Negative",
    "Negative",
    "Negative",
    "Negative"
]
# Split data
X_train, X_test, y_train, y_test = train_test_split(
    texts,
    labels,
    test_size=0.3,
    random_state=42,
    stratify=labels
)
# TF-IDF
vectorizer = TfidfVectorizer()
X_train = vectorizer.fit_transform(X_train)
X_test = vectorizer.transform(X_test)
# Train model
model = MultinomialNB()
model.fit(X_train, y_train)
# Prediction
y_pred = model.predict(X_test)
# Accuracy
accuracy = accuracy_score(y_test, y_pred)
print("Actual Labels:")
print(y_test)
print("\nPredicted Labels:")
print(y_pred)
print("\nAccuracy:", accuracy)
```

```python
#  Classification Model Evaluation

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)
texts = [
    "I love this movie",
    "Excellent movie",
    "Amazing acting",
    "Very good film",
    "Fantastic experience",
    "Great movie",
    "I hate this movie",
    "Terrible movie",
    "Bad acting",
    "Very boring film",
    "Worst experience",
    "Poor movie"
]

labels = [
    "Positive", "Positive", "Positive",
    "Positive", "Positive", "Positive",
    "Negative", "Negative", "Negative",
    "Negative", "Negative", "Negative"
]

# Split dataset
X_train, X_test, y_train, y_test = train_test_split(
    texts,
    labels,
    test_size=0.25,
    random_state=42,
    stratify=labels
)

# TF-IDF feature extraction
vectorizer = TfidfVectorizer()

X_train = vectorizer.fit_transform(X_train)
X_test = vectorizer.transform(X_test)

# Logistic Regression model
model = LogisticRegression()
model.fit(X_train, y_train)

# Prediction
y_pred = model.predict(X_test)
# Evaluation
print("Actual Labels:")
print(y_test)
print("\nPredicted Labels:")
print(y_pred)
print("\nAccuracy:")
print(accuracy_score(y_test, y_pred))
print("\nPrecision:")
print(precision_score(
    y_test, y_pred,
    pos_label="Positive",
    zero_division=0
))
print("\nRecall:")
print(recall_score(
    y_test, y_pred,
    pos_label="Positive",
    zero_division=0
))
print("\nF1-Score:")
print(f1_score(
    y_test, y_pred,
    pos_label="Positive",
    zero_division=0
))
print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))
print("\nClassification Report:")
print(classification_report(
    y_test,
    y_pred,
    zero_division=0
))
```
